## Summary of the Question 
The aim of this project is to create a game called Chocolate Chilli, where a human player competes against a computer. Each player takes turns removing 1, 2 or 3 chocolates from a pile. The player who takes the last chocolate wins, while anyone who starts their turn with no chocolates left loses. As the project progresses, the computer will learn from previous games and improve its chances of winning. For Level 1, the program introduces the game, explains the rules and asks for the player's name and the starting number of chocolates. In Level 2, variables, calculations and methods that return values are introduced to keep track of the chocolates remaining. Level 3 adds decision statements and random computer moves, allowing the game to determine who wins. Level 4 introduces loops so that players can continue taking turns until the game ends, while ensuring that players cannot take more chocolates than are available. For Level 5, arrays and nested loops are introduced to allow multiple games to be played using the same starting number of chocolates. The program records the winner of each game in an array and displays statistics showing how many games the human player and computer have won.

### computerTurn

#### What it does


subtracts the roll to the remaining chocolates 


#### Implementation (how it works)

remaining chocolate is subtracted by the roll then returned. While loop is used to make sure the chocolates remaining subtracted by roll is always 0 and above. if negative it rolls again.  



````java
public static int computerTurn (int chocolatesRemaining)
{
    int roll,temp;
    boolean flag;
    flag = false;
    temp=0;

    
    while (!flag)
    {
        temp =chocolatesRemaining;
        roll=rollDice();
        temp = (temp-roll);
        if (temp >=0)
        {
            System.out.println("Computer Takes "+roll);
            flag = true;
        }    
    }
    return temp;
}
````

### humanTurn

#### What it does


subtracts the roll to the remaining chocolates with 1, 2 or 3 options.


#### Implementation (how it works)

remaining chocolate is subtracted by the human choice of either 1, 2 or 3 then returned, While loop is used so user input is correct with numbers 1-3, and they cannot select a number such that the subtracton is negative. while loop makes sure the user gives a valid input.



````java
public static int humanTurn (int chocolatesRemaining, Scanner input)
{
    boolean flag;
    String amount;
    int intAmount,temp;
    final int MIN_MOVE = 1;
    final int MAX_MOVE = 3;

    flag=false;
    intAmount = 0;
    temp=0;
    
    while (!flag)
    {
        while ((intAmount<MIN_MOVE) || (intAmount>MAX_MOVE))
        {
            System.out.println("how much chocolate you want to take(1-3)");
            amount = input.nextLine();
            intAmount = Integer.parseInt(amount);
            
            if ((intAmount<MIN_MOVE) || (intAmount>MAX_MOVE))
            {
                System.out.println("invalid input,input number between 1 to 3");
            } 
        }
        
        temp = (chocolatesRemaining-intAmount);
        if (temp >= 0)
        {
            System.out.println("Human Takes "+intAmount);
            flag = true;
        }
        else
        {
            System.out.println("invalid input, cannot be negative");
            intAmount = 0;
        }
    }
    return temp;
}
````

### playGames

#### What it does

play multiple games and record wins 
#### Implementation (how it works)

runs a for loop through the array till all games are done and record wins. Runs a while loop to check if game is done and sets computer or human to true if they win. then outputs the winner and stores winner number 1 for human and 2 for computer in array. 



````java
public static void playGames (int numberOfGames, int initialChoc, int[] winners, Scanner input)
{
    boolean flag,human,computer;
    int remainChoc;
    int i;

    for (i=0 ; i<numberOfGames ; i++)
    {
        remainChoc = initialChoc;
    
        System.out.println("");
        System.out.println("You have " + initialChoc + " chocolates to start with.");
        System.out.println("I will go first!");

       flag = false;
       human= false;
       computer =false;
            
       while (!(flag))
       {
           remainChoc = computerTurn(remainChoc);
           System.out.println("there is "+remainChoc+" chocolates left");
           if (checkEndOfGame(remainChoc)) 
           {
              flag =true;
              computer = true;
           }
           else
           {
               remainChoc = humanTurn(remainChoc,input);
               System.out.println("there is "+remainChoc+" chocolates left");
               if (checkEndOfGame(remainChoc))
               {
                   
                   flag = true;
                   human = true;
               }
           }
       }
       if (human)
       {
           System.out.println("human wins");
           winners[i] = 1;
       }
       else if (computer)
       {
           System.out.println("Computer Wins");
           winners[i] = 2;
       }
    }
    return;
}
````

### displayStatistics


#### What it does


calculates and displays the game statistics


#### Implementation (how it works)

runs a for loop till end of array. if number is 1, human wins else if number is 2, computer wins and both are recorded but increasing the wins count. then it displays the win for human and computer.
 


````java
public static void displayStatistics(int[] winners)
{
    int humanWins;
    int computerWins;
    int i;

    humanWins = 0;
    computerWins = 0;

    for (i = 0; i < winners.length; i++)
    {
        if (winners[i] == 1)
        {
            humanWins = humanWins +1;
        }
        else if (winners[i] == 2)
        {
            computerWins = computerWins +1;
        }
    }

    System.out.println("Human wins: " + humanWins);
    System.out.println("Computer wins: " + computerWins);
}
````

````java
// Syed-Nafees Zaman
// 17/11/2026
// VERSION 2
/*The aim of this project is to create a game called Chocolate Chilli, where a human player competes against a computer. 
  Each player takes turns removing 1, 2 or 3 chocolates from a pile. The player who takes the last chocolate wins, 
  while anyone who starts their turn with no chocolates left loses. As the project progresses, 
  the computer will learn from previous games and improve its chances of winning. 
  For Level 1, the program introduces the game, explains the rules and asks for the player's name and the starting number of chocolates. 
  In Level 2, variables, calculations and methods that return values are introduced to keep track of the chocolates remaining. 
  Level 3 adds decision statements and random computer moves, allowing the game to determine who wins. 
  Level 4 introduces loops so that players can continue taking turns until the game ends, while ensuring that players cannot take more chocolates 
  than are available. For Level 5, arrays and nested loops are introduced to allow multiple games to be played using the same starting number of 
  chocolates. The program records the winner of each game in an array and displays statistics showing how many games the human player and computer 
  have won. 
*/

import java.util.Scanner; // Needed to make Scanner available if run outside JHUB
 
class chocolateChilli 
{
    public static void main (String [] a)
    {
        startGame();
        return;
    } // END main

    //display welcome message 
    public static void welcomeMessage() 
    {
        System.out.println("Welcome to Chocolate Chilli!");
    }

    //display rules of game
    public static void explainRules() 
    {
        System.out.println("");
        System.out.println("The rules of Chocolate Chilli are as follows:");
        System.out.println("Players take turns eating 1, 2 or 3 chocolates.");
        System.out.println("Whoever takes the last chocolate wins.");
        System.out.println("The losing player must eat the chilli!");
        System.out.println("");
    }

    //get name of player
    public static String getPlayerName (Scanner input)
    {
      String player1;
      System.out.print("Enter name for Player 1: ");
      player1 = input.nextLine();
      return player1;
    }

    //gets initial chocolates 
    public static int getStartingChocolates (Scanner input)
    {
      String chocolates;
      int intChocolates;
    
      System.out.print("How many chocolates would you like to start with? ");
      chocolates = input.nextLine();
      intChocolates = Integer.parseInt(chocolates);

    return intChocolates;
    }

    //roll between 1-3
    public static int rollDice()
    {
      int randomNum;
      Random random = new Random();

      randomNum = random.nextInt(3) + 1;
      return randomNum;
    }

    //subtracts the roll to the remaining chocolates
    public static int computerTurn (int chocolatesRemaining)
    {
        int roll,temp;
        boolean flag;
        flag = false;
        temp=0;

        while (!flag)
        {
            temp =chocolatesRemaining;
            roll=rollDice();
            temp = (temp-roll);
            if (temp >=0)
            {
                System.out.println("Computer Takes "+roll);
                flag = true;
            }    
        }
        return temp;
    }

    //subtracts the amount the humans takes between 1 to 3  to the remaining chocolates
   public static int humanTurn (int chocolatesRemaining, Scanner input)
   {
       boolean flag;
       String amount;
       int intAmount,temp;
       final int MIN_MOVE = 1;
       final int MAX_MOVE = 3;

       flag=false;
       intAmount = 0;
       temp=0;
    
       while (!flag)
       {
           while ((intAmount<MIN_MOVE) || (intAmount>MAX_MOVE))
           {
               System.out.println("how much chocolate you want to take(1-3)");
               amount = input.nextLine();
               intAmount = Integer.parseInt(amount);
            
               if ((intAmount<MIN_MOVE) || (intAmount>MAX_MOVE))
               {
                   System.out.println("invalid input,input number between 1 to 3");
               } 
           }
        
           temp = (chocolatesRemaining-intAmount);
           if (temp >= 0)
           {
               System.out.println("Human Takes "+intAmount);
               flag = true;
           }
           else
           {
               System.out.println("invalid input, cannot be negative");
               intAmount = 0;
           }
       }
       return temp;
    }

    //Checks for when remaining chocolates are 0 and who won
    public static boolean checkEndOfGame(int chocolatesRemaining)
    {
      final int NO_CHOCOLATES_LEFT = 0;
      if (chocolatesRemaining == NO_CHOCOLATES_LEFT)
      {
        return true;
      }
      else
      {
        return false;
      }
    }

    //asks how many games they want to play
    public static int getNumberOfGames (Scanner input)
    {
        String numberOfGames;
        int intNumberOfGames;

        System.out.print("How many games would you like to play? ");
        numberOfGames = input.nextLine();
        intNumberOfGames = Integer.parseInt(numberOfGames);

        return intNumberOfGames;
    }

    //creates the Winner array depending on the amount of games played
    public static int[] initialiseWinners(int numberOfGames)
    {
        int[] winners = new int[numberOfGames];

        return winners;
    }

    //play multiple games and record wins
    public static void playGames (int numberOfGames, int initialChoc, int[] winners, Scanner input)
    {
        boolean flag,human,computer;
        int remainChoc;
        int i;

        for (i=0 ; i<numberOfGames ; i++)
        {
            remainChoc = initialChoc;
    
            System.out.println("");
            System.out.println("You have " + initialChoc + " chocolates to start with.");
            System.out.println("I will go first!");

            flag = false;
            human= false;
            computer =false;
            
            while (!(flag))
            {
                remainChoc = computerTurn(remainChoc);
                System.out.println("there is "+remainChoc+" chocolates left");
                if (checkEndOfGame(remainChoc)) 
                {
                    flag =true;
                    computer = true;
                }
                else
                {
                    remainChoc = humanTurn(remainChoc,input);
                    System.out.println("there is "+remainChoc+" chocolates left");
                    if (checkEndOfGame(remainChoc))
                    {
                        flag = true;
                        human = true;
                   }
                }
            }
            
            if (human)
            {
                System.out.println("human wins");
                winners[i] = 1;
            }
            else if (computer)
            {
                System.out.println("Computer Wins");
                winners[i] = 2;
            }
        }
        
        return;
    }

    //calculates and displays the game statistics
    public static void displayStatistics(int[] winners)
    {
        int humanWins;
        int computerWins;
        int i;

        humanWins = 0;
        computerWins = 0;

        for (i = 0; i < winners.length; i++)
        {
            if (winners[i] == 1)
            {
                humanWins = humanWins+1;
            }
            else if (winners[i] == 2)
            {
                computerWins = computerWins + 1;
            }
        }

        System.out.println("Human wins: " + humanWins);
        System.out.println("Computer wins: " + computerWins);
    }

    //takes in user name and how many chocolate they start with and required methods from above , then run multiple games till all games finished
    public static void startGame()
    {
        Scanner input = new Scanner(System.in);
        String player;
        int initialChoc,numberOfGames;
        int winners[];

        welcomeMessage();
        explainRules();

        player = getPlayerName(input);
        initialChoc = getStartingChocolates(input);
        numberOfGames = getNumberOfGames(input);
    
        System.out.println("Thanks " + player);
    
        winners = initialiseWinners(numberOfGames);
        playGames(numberOfGames,initialChoc,winners,input);
        displayStatistics(winners);
        
        return;
    }


} // END class chocolateChilli
````
